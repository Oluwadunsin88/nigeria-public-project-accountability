```python

import pandas as pd

# Load the public project dataset
file_path = "/content/solution-lens-projects-2026-10-26 (1).csv"
df = pd.read_csv(file_path)

# Display the first five records
display(df.head())
```

```python

# Check the dataset size
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Display all column names
print("\nColumn names:")
print(df.columns.tolist())

# Count missing values in each column
print("\nMissing values:")
print(df.isnull().sum().sort_values(ascending=False))
```

```python

# Calculate missing values and percentages
missing_summary = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing Percentage": (df.isnull().mean() * 100).round(2),
    "Available Count": df.notnull().sum()
})

missing_summary = missing_summary.sort_values(
    by="Missing Percentage",
    ascending=False
)

display(missing_summary)
```

```python

# 1. Inspect data types and available values
print("DATA TYPES AND AVAILABLE VALUES")
df.info()

# 2. Check reported progress values
print("\nPROGRESS VALUES")
print(df["Progress (%)"].describe())

# 3. Check project status categories
print("\nPROJECT STATUS CATEGORIES")
print(df["Status"].value_counts(dropna=False))

# 4. Check duplicate project titles
print("\nDUPLICATE PROJECT TITLES")
print("Number of repeated titles:",
      df["Title"].duplicated(keep=False).sum())

# 5. Inspect schedule values
print("\nSCHEDULE CATEGORIES")
print(df["Schedule"].value_counts(dropna=False))

# 6. Inspect project counts by LGA
print("\nPROJECTS BY LGA")
print(df["LGA"].value_counts(dropna=False))
```

```python

# Select projects that have reported progress
progress_df = df[df["Progress (%)"].notna()].copy()

# Compare reported progress with project status
print("PROGRESS BY PROJECT STATUS")

display(
    progress_df.groupby("Status")["Progress (%)"].agg(
        ["count", "mean", "median", "min", "max"]
    ).round(2)
)

# Find projects whose progress may need investigation
print("\nPROJECTS REQUIRING STATUS REVIEW")

review_df = df[
    ((df["Status"] == "Completed") &
     (df["Progress (%)"] < 100)) |
    ((df["Status"] == "In progress") &
     (df["Progress (%)"] == 100))
]

display(
    review_df[
        ["Code", "Title", "Status", "LGA", "Progress (%)"]
    ]
)

print("\nNumber of projects flagged for review:",
      len(review_df))

# List all projects with reported progress
print("\nALL PROJECTS WITH REPORTED PROGRESS")

display(
    progress_df[
        ["Code", "Title", "Status", "LGA", "Progress (%)"]
    ].sort_values("Progress (%)", ascending=True)
)
```

```python

# Create an audit copy; preserve the original dataset
audit_df = df.copy()

# Create a list of review flags for each project
def identify_issues(row):
    issues = []

    if pd.isna(row["LGA"]):
        issues.append("Missing LGA")

    if pd.isna(row["Progress (%)"]):
        issues.append("Missing progress")

    if pd.isna(row["Responsible MDA"]):
        issues.append("Missing responsible MDA")

    if pd.isna(row["Schedule"]):
        issues.append("Missing schedule")

    if row["Status"] == "In progress" and pd.isna(row["Last update"]):
        issues.append("Missing last update")

    if pd.notna(row["Progress (%)"]):
        if not 0 <= row["Progress (%)"] <= 100:
            issues.append("Progress outside 0-100 range")

        if row["Status"] == "In progress" and row["Progress (%)"] == 100:
            issues.append("Verify status: 100% progress")

        if row["Status"] == "Completed" and row["Progress (%)"] < 100:
            issues.append("Verify status: progress below 100%")

    return "; ".join(issues)

# Apply the checks
audit_df["Review Flags"] = audit_df.apply(identify_issues, axis=1)

# Separate records requiring review
flagged_projects = audit_df[
    audit_df["Review Flags"] != ""
].copy()

# Display the summary
print("Total projects:", len(audit_df))
print("Projects flagged for review:", len(flagged_projects))
print("Projects without these review flags:",
      len(audit_df) - len(flagged_projects))

# View the flagged records
display(
    flagged_projects[
        ["Code", "Title", "Status", "LGA",
         "Progress (%)", "Review Flags"]
    ].head(20)
)
```
